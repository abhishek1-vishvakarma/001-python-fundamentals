```python
class BankManagement():
  def __init__(self, amount):
    self.amount = amount

  def currentBalance(self):
        return self.amount

  def deposite(self, amount):
    self.amount += amount
    return self.amount
  
  def withdraw(self, amount):
    if self.amount < amount:
      print("Insufficient balance")
    else:
      self.amount -= amount
      return self.amount
  
res = BankManagement(100)
res.deposite(100)
print(res.currentBalance())
res.withdraw(20)
print(res.currentBalance())

```

```python
def maximumNum(arr1):
  obj = []
  for e in range(len(arr1)):
   for l in range(len(arr1)):
    if arr1[e] < arr1[l]:
      temp = arr1[e]
      arr1[e] = arr1[l]
      arr1[l] = temp

  obj.append(arr1[:1])
  obj.append(arr1[-1::])
  return obj

arr1 = [20, 10, 80, 40, 50, 60]
res = maximumNum(arr1)
print(res)
```

```python
def duplicateRemove(arr1):
  result = []
  for e in arr1:
    if e not in result:
      result.append(e)
  return result

print(duplicateRemove([20, 10, 10, 80, 40, 50, 60, 60]))
```
